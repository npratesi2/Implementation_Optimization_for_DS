# Dry Bean Dataset: missing values check and stratified sampling

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

SEED, N = 0, 1000
df = pd.read_excel('../Dataset/Dry_Bean_Dataset.xlsx')
print(df.shape)
print('Missing values:', df.isna().sum().sum())
```

## Stratified sampling (N = 1000) and standardization

```python
assert df.isna().sum().sum() == 0, 'Dataset has missing values'

sample, _ = train_test_split(df, train_size=N, stratify=df['Class'], random_state=SEED)
sample = sample.reset_index(drop=True)

features = sample.columns.drop('Class')
sample[features] = StandardScaler().fit_transform(sample[features])
```

```python
sample.to_csv('../Dataset/dry_bean_sample.csv', index=False)
```
